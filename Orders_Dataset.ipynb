```python
city_map = {
    # ALL CAPS entries where title() alone isn't quite right
    'RENALAKHURD': 'Renala Khurd',        # missing space, title() would give "Renalakhurd"

    # mixed-case entries title() won't touch (not fully upper or fully lower)
    'Mingora (swat)': 'Mingora (Swat)',
    'Mirpur (a. K)': 'Mirpur (AK)',
    'Nowshera virkan': 'Nowshera Virkan',
    'Pindi bhattian': 'Pindi Bhattian',
    'Rahim yar khan': 'Rahim Yar Khan',
    'Tando ala yar': 'Tando Allahyar',
    'Dera ala yar': 'Dera Allah Yar',

    # genuine misspellings
    'Toba Tek Sing': 'Toba Tek Singh',
    'Kark': 'Karak',

    # stray punctuation
    'Shorkot Cantt.': 'Shorkot Cantt',

    # duplicate/redundant variant of an existing city
    'Batkhela Distt Malakand': 'Batkhela',

    # non-Latin script entry, transliterated for consistency
    'کھڈیاں خاص': 'Khadian Khas',
}

df['city'] = df['city'].replace(city_map)
```

```python
df['city'] = df['city'].apply(
    lambda c: c.title() if (c.isupper() or c.islower()) else c
)
```

```python
# get the "typical" single-unit price for each SKU from quantity=1 rows
unit_price = df[df['quantity'] == 1].groupby('sku')['sales'].median()

# for quantity>1 rows, compare actual sales to what it *should* be
check = df[df['quantity'] > 1].copy()
check['expected_price'] = check['sku'].map(unit_price)
check['expected_sales'] = check['expected_price'] * check['quantity']
check['mismatch'] = check['sales'] != check['expected_sales']

print(check[check['mismatch']][['sku','quantity','sales','expected_sales']])
```

```python
check['sales_flag'] = np.where(
    (check['sales'] == check['expected_price']) &
    (check['quantity'] > 1),
    'quantity not applied - verify',
    'OK'
)
```

```python
check[check['sales_flag'] == 'quantity not applied - verify']
```

```python
df['sales_flag'] = 'OK'

df.loc[check.index, 'sales_flag'] = check['sales_flag']
```
